```python
file_path = 'data/namesbystate_ca.txt.gz'
column_labels = ['State', 'Sex', 'Year', 'Name', 'Count']

babynames = pd.read_csv(file_path, names=column_labels)
```

```python
len(babynames)
```

```python
babynames.head()
```

### Q4a 用`.loc`选择`Name`和`Year`列

```python
name_and_year = babynames.loc[:,['Name', 'Year']]
name_and_year[:5]
```

```python
grader.check("q4a")
```

### Q4b 用`[]`完成同样的选择

```python
name_and_year = babynames[['Name', 'Year']]
name_and_year[:5]
```

### Q5 使用布尔数组，选择2000年计数大于3000的名字；可使用`[]`或`.loc`完成

```python
result = babynames[(babynames['Year'] == 2000) & (babynames['Count'] > 3000)]
result.head()
```

```python
grader.check("q5")
```

### Q6a 使用`.str.get`在“babynames”中添加一个名为“First Letter”的列，其中包含每个婴儿名字的第一个字母

```python
babynames['First Letter'] = babynames['Name'].str.get(0)
babynames
```

```python
grader.check("q6a")
```

### Q6b 计算2022年，有多少名字是以A开头的

```python
babynames_2022 = babynames[babynames['Year'] == 2022]
just_A_names_2022 = babynames_2022[babynames_2022['First Letter'] == 'A']
number_A_names = just_A_names_2022['Count'].sum()
number_A_names
```

### 使用列表和列名创建dataframe
pd.DataFrame(列表，列名columns)

```python
pd.DataFrame([1, 2, 3], columns=['Numbers'])
```

**传递多个列表的形式实际是以行的形式，一个列表表示一行，列表长度==序列数，列表个数==行数**

```python
pd.DataFrame([[1, 'a'], [2, 'b'], [3, 'c']], columns=['Numbers', 'Letters'])
```

### 使用字典创建dataframe
{"键1"：[值1、、、], "键2" : [值2、、、]}
键数是列数，值数是行数


```python
pd.DataFrame({"Fruit" : ["Strawberry", "Orange", "Peach"],
              "Price": [5.49, 3.99, 1.99]
              })
```

### 使用Series创建dataframe
先创建多个series：pd.Series(列表，索引index)<br>
将多个series合并：pd.DataFrame({"列名1":series1, "列名2":series2})

```python
s_a = pd.Series(["a1", "a2", "a3"], index=["r1", "r2", "r3"])
s_b = pd.Series(["b1", "b2", "b3"], index=["r1", "r2", "r3"])
```

```python
pd.DataFrame({"A-column": s_a, "B-column": s_b})
```

## dataframe的loc和iloc
**loc和iloc都是用来获取dataframe的行和列的**<br>
df.loc[行索引, 列索引]：使用行和列的**标签**来获取数据，<br>
df.iloc[行索引, 列索引]：使用行和列的**整数位置**来获取数据<br>
loc可以只提供行，只提供列时的格式：loc[:,'']；<br>
iloc同理<br>
### **索引可以用列表、切片、单独值共三种方式提供**
iloc，行索引不受index设置的影响，列索引时0是从index列的后一列开始算<br>

```python
elections.loc['Andrew Jackson', 'Party']
```

```python
elections.loc[['Andrew Jackson', 'Jill Stein'], 'Party':'%']
```

```python
elections.loc[['Andrew Jackson', 'Jill Stein'], 'Party']
```

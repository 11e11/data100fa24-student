#### Q2b 进一步检查ins['score']<br>
ins['Missing Score']列存储了ins['score']中是否为缺失值（-1）。<br>
使用groupby，传入两个参数：'type'和'Missing Score'，<br>
然后使用聚合函数agg，对分组后数据的score进行count聚合<br>
然后使用rename将count列重命名为Count。<br>

```python
ins['score'].value_counts().head()
```

```python
ins['Missing Score'] = (ins['score'] == -1).astype("str")
# ins_missing_score_group = ins.groupby(['type', 'Missing Score'])['score'].agg(['count']).rename(columns={'count': 'Count'})
ins_missing_score_group = ins.groupby(['type', 'Missing Score']).agg(Count = ('score', 'count'))
ins_missing_score_group
```

```python
grader.check("q1b")
```

#### Q1c 用另一种方法（pivot_table）得到'Missing Score'和'type'的组合<br>
使用pivot_table方法，创建一个名为'ins_missing_score_pivot'的DataFrame，<br>
其中索引为'type'，列为'Missing Score'，值为'score'的计数。<br>
请注意，pivot_table方法的aggfunc参数默认为'mean'，但我们需要使用'count'。<br>
请确保填充缺失值为0<br>
最后，添加一列'Total'，该列是每行的总和，并按'Total'列降序排序。<br>

```python
ins['Missing Score'] = (ins['score'] == -1).astype("str")
ins_missing_score_pivot = ins.pivot_table(values='score', index='type', columns='Missing Score', aggfunc='count', fill_value=0)

ins_missing_score_pivot['Total'] = ins_missing_score_pivot.sum(axis=1)
ins_missing_score_pivot = ins_missing_score_pivot.sort_values(by='Total', ascending=False)

ins_missing_score_pivot
```

```python
grader.check("q1c")
```

### Q3 找到最好的甜品店（蛋糕店）

用.str.contains('cake')查看是否包含cake'<br>

#### Q3b 找到至少经过3次检查的蛋糕店cake_at_least_3DataFrame<br>
① 先用value_counts()对bid出现次数排序，用lambda表达式选中出现次数≥三次的<br>
 用isin()方法筛选出这些bid对应的行<br>
② 用groupby()方法对bid分组，使用filter()方法和lambda表达式筛选

```python
# cake_at_least_3 = cake_shops[cake_shops['bid'].isin(cake_shops['bid'].value_counts()[lambda x : x >= 3].index)]
cake_at_least_3 = cake_shops.groupby('bid').filter(lambda x: len(x) >= 3)
cake_at_least_3.head()
```

## Q3d 按照蛋糕店名称中元音字母的数量对cake_at_least_3进行降序排序<br>
返回排序后前三个名称**唯一**的蛋糕店<br>
名称唯一：drop_duplicates()方法<br>

```python
def count_vowels(name):
    vowels = 'aeiouAEIOU'
    return sum([letter in vowels for letter in name])
```

```python
sorted_by_vowel_count = cake_at_least_3['name'].apply(count_vowels).sort_values(ascending=False)
# sorted_by_vowel_count
top_cakes = cake_at_least_3.loc[sorted_by_vowel_count.index, 'name'].drop_duplicates().head(3).tolist()

top_cakes
```

### Q4 找到具有三次及以上评分记录的餐厅的历史最高分和最低分的最大差值的餐厅max_swing<br>

```python
# 找到具有三次及以上评分记录的餐厅
# shops_at_least_3 = ins_named[ins_named['bid'].isin(ins_named['bid'].value_counts()[lambda x: x >= 3].index)]

shops_at_least_3 = ins_named.groupby('bid').filter(lambda x: len(x) >= 3)

# 按照bid分组，计算每个餐厅的最高分和最低分
# 并获取餐厅名称
# 注意：这里使用agg函数来同时计算多个聚合值，得到的是多级索引的dataframe，所以后面重新赋值了columns
score_swing = shops_at_least_3.groupby('bid').agg({'score':['max', 'min'],'name':'first'})
score_swing.columns = ['max', 'min', 'name']
# 计算差值swing
score_swing['swing'] = score_swing['max'] - score_swing['min']

# 找到最大差值的索引idxmax，用loc定位到行索引和列标签'name'
# max_swing = score_swing.loc[score_swing['swing'].idxmax(), 'name']
max_swing = score_swing.sort_values(by='swing').iloc[0]['name']
```
